```
!pip install gensim
!pip install nltk
!pip install tabulate
import bs4 as bs
import urllib.request
import re
import nltk
nltk.download('punkt')
nltk.download('stopwords')

from tabulate import tabulate
```

```

scrapped_data = urllib.request.urlopen('https://en.wikipedia.org/wiki/Artificial_intelligence')
article = scrapped_data .read()
#print(article)
parsed_article = bs.BeautifulSoup(article,'lxml')
#print(parsed_article.prettify())
paragraphs = parsed_article.find_all('p')

article_text = ""

for p in paragraphs:
    article_text += p.text
print(article_text)
```

```
# Cleaing the text
processed_article = article_text.lower()
processed_article = re.sub('[^a-zA-Z]', ' ', processed_article )
processed_article = re.sub(r'\s+', ' ', processed_article)

# Preparing the dataset
all_sentences = nltk.sent_tokenize(processed_article)
#print(all_sentences)
all_words = [nltk.word_tokenize(sent) for sent in all_sentences]
print(all_words)
# Removing Stop Words
from nltk.corpus import stopwords
for i in range(len(all_words)):
    all_words[i] = [w for w in all_words[i] if w not in stopwords.words('english')]
print(all_words)
```

```
# from gensim.models import Word2Vec

# word2vec = Word2Vec(all_words, min_count=2)
```

```
from gensim.models import Word2Vec

# Train the Word2Vec model
word2vec = Word2Vec(sentences=all_words, min_count=2, vector_size=100)

# Access the vocabulary using the updated attributes in Gensim 4.x
vocabulary = word2vec.wv.key_to_index

# Print the vocabulary (word to index mapping)
print(vocabulary)
```

```
v1 = word2vec.wv['artificial']
print(v1)
```

```
v1 = word2vec.wv['ai']
print(v1)
```

```

#sim_words = word2vec.wv.most_similar('artificial')
sim_words = word2vec.wv.most_similar(positive=['ai', 'learning', 'probability'], topn=13)
print(tabulate(sim_words))
```
